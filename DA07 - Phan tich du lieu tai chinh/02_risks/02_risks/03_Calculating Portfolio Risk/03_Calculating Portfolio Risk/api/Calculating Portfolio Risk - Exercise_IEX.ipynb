## Calculating Portfolio Risk

Calculate the risk of an equally weighted portfolio composed of Microsoft and Apple. The data can be obtained from IEX for the period 1st of January 2007 until today. 

*Hint: The code we went through in the lecture is what you need here. You will need to import the data first. The previous lessons could be a good reference point for that! :) *

```python
import numpy as np
import pandas as pd
from pandas_datareader import data as wb
import matplotlib.pyplot as plt
import yfinance as yf
```

```python
tickers = ['MSFT', 'AAPL']

sec_data = pd.DataFrame()

for t in tickers:
    sec_data[t] = yf.download(t, start='2015-1-1')['Close']
```

```python
sec_data.tail()
```

```python
sec_returns = np.log(sec_data / sec_data.shift(1))
```

## Covariance and Correlation


\begin{eqnarray*}
Covariance Matrix: \  \   
\Sigma = \begin{bmatrix}
        \sigma_{1}^2 \ \sigma_{12} \ \dots \ \sigma_{1I} \\
        \sigma_{21} \ \sigma_{2}^2 \ \dots \ \sigma_{2I} \\
        \vdots \ \vdots \ \ddots \ \vdots \\
        \sigma_{I1} \ \sigma_{I2} \ \dots \ \sigma_{I}^2
    \end{bmatrix}
\end{eqnarray*}

*****

Covariance matrix:

```python
cov_matrix = sec_returns.cov()
cov_matrix
```

***

Correlation matrix:

```python
cov_matrix_a = sec_returns.cov() * 250
cov_matrix_a
```

```python
corr_matrix = sec_returns.corr()
corr_matrix
```

## Calculating Portfolio Risk

Equal weigthing scheme:

```python
weights = np.array([0.5, 0.5])
```

Portfolio Variance:

```python
pfolio_var = np.dot(weights.T, np.dot(sec_returns.cov() * 250, weights))
pfolio_var
```

Portfolio Volatility:

```python
pfolio_vol = (np.dot(weights.T, np.dot(sec_returns.cov() * 250, weights))) ** 0.5
pfolio_vol
```

```python
print (str(round(pfolio_vol, 5) * 100) + ' %')
```
