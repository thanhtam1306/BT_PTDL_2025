### Simple Rate of Return

Calculate the simple returns of ‘MSFT’ for the given timeframe.

$$
\frac{P_1 - P_0}{P_0} = \frac{P_1}{P_0} - 1
$$

```python
PG['simple_return'] = (PG['Close'] / PG['Close'].shift(1)) - 1
print (PG['simple_return'])
```
