```python
# train a linear regrssion model using Pseudo inverse of A

import numpy as np

A = [[40, 30, 1],
     [30, 80, 1],
     [20, 70, 1],
     [70, 40, 1],
     [60, 20, 1]]

Z = [[1], [-1], [-1], [1], [1]]
```

```python
At = np.transpose(A)

print(At)
```

```python
AtA = np.dot(At,A)
AtZ = np.dot(At,Z)

print(AtA)
print(AtZ)
```

```python
inv_AtA = np.linalg.inv(AtA)
X = np.dot(inv_AtA,AtZ)
```

```python
print(X)
```

### Hồi quy tuyến tính

```python
import numpy as np

# Dữ liệu
X = np.array([[40, 30, 1],
              [30, 80, 1],
              [20, 70, 1],
              [70, 40, 1],
              [60, 20, 1]])
Y = np.array([[1], [-1], [-1], [1], [1]])

# Tính theta bằng pseudo-inverse
Xt = X.T
theta = np.linalg.inv(Xt @ X) @ Xt @ Y

print("Weights (theta):\n", theta)

# Dự đoán
Y_pred = X @ theta
print("Predictions:\n", Y_pred)

# Residual error
residual = Y - Y_pred
print("Residual:\n", residual)
```

### Hồi quy logistic

```python
import numpy as np

# Dữ liệu (giả sử Y là 0 hoặc 1)
X = np.array([[40, 30, 1],
              [30, 80, 1],
              [20, 70, 1],
              [70, 40, 1],
              [60, 20, 1]])
Y = np.array([[1], [0], [0], [1], [1]])

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Khởi tạo weights
theta = np.zeros((X.shape[1],1))
alpha = 0.001
iterations = 10000

# Gradient descent
for i in range(iterations):
    z = X @ theta
    h = sigmoid(z)
    gradient = X.T @ (h - Y) / Y.size
    theta -= alpha * gradient

print("Weights (theta):\n", theta)

# Dự đoán xác suất
Y_prob = sigmoid(X @ theta)
print("Predicted probabilities:\n", Y_prob)

# Chuyển xác suất sang nhãn 0/1
Y_pred = (Y_prob >= 0.5).astype(int)
print("Predicted labels:\n", Y_pred)
```
